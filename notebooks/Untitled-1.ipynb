```python
# ============================================================
# COVID-19 END-TO-END DATA PIPELINE
# GitHub -> Python -> Cleaning -> KPIs -> CSV -> SQLite Database
# ============================================================

# -----------------------------
# 1. IMPORT LIBRARIES
# -----------------------------
import requests
import pandas as pd
import numpy as np
import sqlite3
from pathlib import Path
from datetime import datetime
from io import StringIO
import time

print("STEP 1: Libraries imported successfully")


# -----------------------------
# 2. CREATE PROJECT FOLDERS
# -----------------------------
current_path = Path.cwd()

# If notebook is running from /notebooks, use parent as project root
if current_path.name.lower() == "notebooks":
    project_root = current_path.parent
else:
    project_root = current_path

data_folder = project_root / "data"
raw_folder = data_folder / "raw"
processed_folder = data_folder / "processed"
database_folder = project_root / "database"
dashboard_folder = project_root / "dashboard"

raw_folder.mkdir(parents=True, exist_ok=True)
processed_folder.mkdir(parents=True, exist_ok=True)
database_folder.mkdir(parents=True, exist_ok=True)
dashboard_folder.mkdir(parents=True, exist_ok=True)

print("STEP 2: Project folders created")
print("Project Root:", project_root)


# -----------------------------
# 3. GET COMPLETE GITHUB FILE LIST
# -----------------------------
# We use GitHub Tree API here so we can discover the complete
# repository file structure in one request.

tree_api_url = (
    "https://api.github.com/repos/"
    "CSSEGISandData/COVID-19/git/trees/master?recursive=1"
)

response = requests.get(tree_api_url, timeout=60)
response.raise_for_status()

tree_data = response.json()

if tree_data.get("truncated"):
    raise Exception(
        "GitHub returned a truncated repository tree. "
        "Please stop and review before continuing."
    )

print("STEP 3: GitHub repository accessed successfully")
print("GitHub Status Code:", response.status_code)


# -----------------------------
# 4. FIND DAILY REPORT CSV FILES
# -----------------------------
daily_report_folder = (
    "csse_covid_19_data/"
    "csse_covid_19_daily_reports/"
)

csv_paths = []

for item in tree_data["tree"]:
    path = item.get("path", "")

    if (
        item.get("type") == "blob"
        and path.startswith(daily_report_folder)
        and path.endswith(".csv")
    ):
        csv_paths.append(path)

print("STEP 4: Daily report files discovered")
print("Total CSV files found:", len(csv_paths))


# -----------------------------
# 5. KEEP ONLY VALID DATE FILES
# -----------------------------
valid_files = []

for path in csv_paths:
    filename = Path(path).name

    try:
        report_date = datetime.strptime(
            filename,
            "%m-%d-%Y.csv"
        )

        valid_files.append(
            {
                "path": path,
                "filename": filename,
                "report_date": report_date
            }
        )

    except ValueError:
        # Ignore non-date CSV files if any
        pass

valid_files = sorted(
    valid_files,
    key=lambda x: x["report_date"]
)

print("STEP 5: Valid dated CSV files found:", len(valid_files))

if len(valid_files) == 0:
    raise Exception("No dated COVID CSV files were found.")


print(
    "Date Range:",
    valid_files[0]["report_date"].date(),
    "to",
    valid_files[-1]["report_date"].date()
)


# -----------------------------
# 6. DOWNLOAD + CLEAN + AGGREGATE
# -----------------------------
# Instead of keeping millions of province/county rows,
# we create a country-day analytical dataset.
#
# This keeps the dashboard dataset smaller and easier to query.

session = requests.Session()

all_daily_country_data = []
failed_files = []

rename_columns = {
    "Province/State": "Province_State",
    "Country/Region": "Country_Region",
    "Last Update": "Last_Update",
    "Latitude": "Lat",
    "Longitude": "Long_"
}

# Known historical naming differences
country_aliases = {
    "Mainland China": "China",
    "Republic of Korea": "Korea, South",
    "Iran (Islamic Republic of)": "Iran",
    "Viet Nam": "Vietnam",
    "Russian Federation": "Russia"
}


for index, file_info in enumerate(valid_files, start=1):

    path = file_info["path"]
    report_date = file_info["report_date"]

    raw_url = (
        "https://raw.githubusercontent.com/"
        "CSSEGISandData/COVID-19/master/"
        + path
    )

    try:
        file_response = session.get(
            raw_url,
            timeout=60
        )

        file_response.raise_for_status()

        daily_df = pd.read_csv(
            StringIO(file_response.text),
            low_memory=False
        )

        # Standardize changing column names
        daily_df = daily_df.rename(
            columns=rename_columns
        )

        # Required geographical column
        if "Country_Region" not in daily_df.columns:
            failed_files.append(
                (file_info["filename"], "Country column missing")
            )
            continue

        # Standardize country text
        daily_df["Country_Region"] = (
            daily_df["Country_Region"]
            .astype(str)
            .str.strip()
            .replace(country_aliases)
        )

        # Convert KPI columns to numeric
        metrics = [
            "Confirmed",
            "Deaths",
            "Recovered"
        ]

        for metric in metrics:
            if metric not in daily_df.columns:
                daily_df[metric] = np.nan

            daily_df[metric] = pd.to_numeric(
                daily_df[metric],
                errors="coerce"
            )

        # Aggregate province/county records to country level
        country_daily = (
            daily_df
            .groupby(
                "Country_Region",
                as_index=False
            )
            .agg(
                Confirmed=(
                    "Confirmed",
                    lambda x: x.sum(min_count=1)
                ),
                Deaths=(
                    "Deaths",
                    lambda x: x.sum(min_count=1)
                ),
                Recovered=(
                    "Recovered",
                    lambda x: x.sum(min_count=1)
                )
            )
        )

        country_daily["Report_Date"] = pd.Timestamp(
            report_date.date()
        )

        country_daily["Source_File"] = file_info[
            "filename"
        ]

        all_daily_country_data.append(
            country_daily
        )

    except Exception as e:
        failed_files.append(
            (file_info["filename"], str(e))
        )

    # Progress
    if index % 50 == 0 or index == len(valid_files):
        print(
            f"Processed {index}/{len(valid_files)} files"
        )


if len(all_daily_country_data) == 0:
    raise Exception(
        "No CSV files could be processed."
    )

print("STEP 6: CSV extraction completed")
print("Failed files:", len(failed_files))


# -----------------------------
# 7. COMBINE ALL DAILY DATA
# -----------------------------
covid = pd.concat(
    all_daily_country_data,
    ignore_index=True
)

print("STEP 7: All daily reports combined")
print("Rows before final cleaning:", len(covid))


# -----------------------------
# 8. FINAL DATA CLEANING
# -----------------------------
covid["Report_Date"] = pd.to_datetime(
    covid["Report_Date"]
)

covid["Country_Region"] = (
    covid["Country_Region"]
    .astype(str)
    .str.strip()
)

numeric_columns = [
    "Confirmed",
    "Deaths",
    "Recovered"
]

for column in numeric_columns:
    covid[column] = pd.to_numeric(
        covid[column],
        errors="coerce"
    )

# Remove invalid country labels
covid = covid[
    covid["Country_Region"].notna()
]

covid = covid[
    covid["Country_Region"] != ""
]

covid = covid[
    covid["Country_Region"].str.lower() != "nan"
]

# One country + one date should only appear once
covid = (
    covid
    .sort_values(
        ["Country_Region", "Report_Date"]
    )
    .drop_duplicates(
        subset=[
            "Country_Region",
            "Report_Date"
        ],
        keep="last"
    )
    .reset_index(drop=True)
)

print("STEP 8: Cleaning completed")
print("Clean rows:", len(covid))


# -----------------------------
# 9. CREATE CALCULATED METRICS
# -----------------------------
covid = covid.sort_values(
    ["Country_Region", "Report_Date"]
)

# Daily changes from cumulative totals
covid["New_Cases"] = (
    covid
    .groupby("Country_Region")["Confirmed"]
    .diff()
)

covid["New_Deaths"] = (
    covid
    .groupby("Country_Region")["Deaths"]
    .diff()
)

# Sometimes cumulative datasets are corrected downward.
# Keep the original calculated difference, but also create
# dashboard-safe non-negative measures.

covid["New_Cases_Adjusted"] = (
    covid["New_Cases"]
    .clip(lower=0)
)

covid["New_Deaths_Adjusted"] = (
    covid["New_Deaths"]
    .clip(lower=0)
)

# Case Fatality Rate
covid["Case_Fatality_Rate"] = np.where(
    covid["Confirmed"] > 0,
    (covid["Deaths"] / covid["Confirmed"]) * 100,
    np.nan
)

print("STEP 9: Calculated KPI fields created")


# -----------------------------
# 10. DATA QUALITY CHECKS
# -----------------------------
duplicate_count = covid.duplicated(
    subset=[
        "Country_Region",
        "Report_Date"
    ]
).sum()

quality_report = pd.DataFrame(
    {
        "Check": [
            "Total Rows",
            "Unique Countries",
            "Minimum Date",
            "Maximum Date",
            "Duplicate Country-Date Records",
            "Missing Confirmed",
            "Missing Deaths",
            "Missing Recovered",
            "Failed Source Files"
        ],
        "Value": [
            len(covid),
            covid["Country_Region"].nunique(),
            covid["Report_Date"].min(),
            covid["Report_Date"].max(),
            duplicate_count,
            covid["Confirmed"].isna().sum(),
            covid["Deaths"].isna().sum(),
            covid["Recovered"].isna().sum(),
            len(failed_files)
        ]
    }
)

print("STEP 10: Data-quality report created")


# -----------------------------
# 11. GLOBAL DAILY KPIs
# -----------------------------
global_daily = (
    covid
    .groupby(
        "Report_Date",
        as_index=False
    )
    .agg(
        Total_Confirmed=(
            "Confirmed",
            lambda x: x.sum(min_count=1)
        ),
        Total_Deaths=(
            "Deaths",
            lambda x: x.sum(min_count=1)
        ),
        Total_Recovered=(
            "Recovered",
            lambda x: x.sum(min_count=1)
        )
    )
    .sort_values("Report_Date")
)

global_daily["New_Global_Cases"] = (
    global_daily["Total_Confirmed"]
    .diff()
)

global_daily["New_Global_Deaths"] = (
    global_daily["Total_Deaths"]
    .diff()
)

global_daily["Global_CFR"] = np.where(
    global_daily["Total_Confirmed"] > 0,
    (
        global_daily["Total_Deaths"]
        / global_daily["Total_Confirmed"]
    ) * 100,
    np.nan
)

print("STEP 11: Global daily KPIs created")


# -----------------------------
# 12. LATEST COUNTRY SNAPSHOT
# -----------------------------
latest_date = covid["Report_Date"].max()

latest_country = (
    covid[
        covid["Report_Date"] == latest_date
    ]
    .sort_values(
        "Confirmed",
        ascending=False
    )
    .reset_index(drop=True)
)

print("STEP 12: Latest country snapshot created")
print("Latest reporting date:", latest_date.date())


# -----------------------------
# 13. SAVE CLEAN CSV FILES
# -----------------------------
clean_csv_path = (
    processed_folder
    / "covid_country_daily.csv"
)

global_csv_path = (
    processed_folder
    / "global_daily_kpis.csv"
)

latest_csv_path = (
    processed_folder
    / "latest_country_snapshot.csv"
)

quality_csv_path = (
    processed_folder
    / "data_quality_report.csv"
)

covid.to_csv(
    clean_csv_path,
    index=False
)

global_daily.to_csv(
    global_csv_path,
    index=False
)

latest_country.to_csv(
    latest_csv_path,
    index=False
)

quality_report.to_csv(
    quality_csv_path,
    index=False
)

print("STEP 13: CSV outputs saved")


# -----------------------------
# 14. SAVE TO SQLITE DATABASE
# -----------------------------
database_path = (
    database_folder
    / "covid_analysis.db"
)

connection = sqlite3.connect(
    database_path
)

covid.to_sql(
    "covid_country_daily",
    connection,
    if_exists="replace",
    index=False
)

global_daily.to_sql(
    "global_daily_kpis",
    connection,
    if_exists="replace",
    index=False
)

latest_country.to_sql(
    "latest_country_snapshot",
    connection,
    if_exists="replace",
    index=False
)

quality_report.to_sql(
    "data_quality_report",
    connection,
    if_exists="replace",
    index=False
)

connection.close()

print("STEP 14: SQLite database created")


# -----------------------------
# 15. FINAL KPI SUMMARY
# -----------------------------
latest_global = global_daily.iloc[-1]

print("\n")
print("=" * 60)
print("COVID-19 PIPELINE COMPLETED SUCCESSFULLY")
print("=" * 60)

print(
    "Date Range:",
    covid["Report_Date"].min().date(),
    "to",
    covid["Report_Date"].max().date()
)

print(
    "Countries:",
    covid["Country_Region"].nunique()
)

print(
    "Analytical Rows:",
    len(covid)
)

print(
    "Latest Total Confirmed:",
    f"{latest_global['Total_Confirmed']:,.0f}"
)

print(
    "Latest Total Deaths:",
    f"{latest_global['Total_Deaths']:,.0f}"
)

print(
    "Latest Global CFR:",
    f"{latest_global['Global_CFR']:.2f}%"
)

print("\nOUTPUT FILES:")
print("Clean Dataset:", clean_csv_path)
print("Global KPIs:", global_csv_path)
print("Latest Country Data:", latest_csv_path)
print("Quality Report:", quality_csv_path)
print("SQLite Database:", database_path)

print("=" * 60)


# -----------------------------
# 16. DISPLAY DATA QUALITY REPORT
# -----------------------------
display(quality_report)


# -----------------------------
# 17. DISPLAY TOP 10 COUNTRIES
# -----------------------------
display(
    latest_country[
        [
            "Country_Region",
            "Confirmed",
            "Deaths",
            "Case_Fatality_Rate"
        ]
    ].head(10)
)
```
